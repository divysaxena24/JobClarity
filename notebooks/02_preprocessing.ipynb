```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/fake_job_postings.csv")

df.head()
```

```python
df = df.drop(columns=["job_id", "salary_range"])
```

```python
text_columns = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

df[text_columns] = df[text_columns].fillna("")
```

```python
categorical_columns = [
    "location",
    "department",
    "employment_type",
    "required_experience",
    "required_education",
    "industry",
    "function"
]

df[categorical_columns] = df[categorical_columns].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.to_csv("../data/processed/processed_jobs.csv", index=False)
```
